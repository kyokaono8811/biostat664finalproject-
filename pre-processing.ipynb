```python
import pandas as pd
from pathlib import Path

# Folder containing the patient files
data_dir = Path.home() / "664" / "physionet2019" / "training" / "training_setA"

# Read one patient's file
df = pd.read_csv(data_dir / "p000001.psv", sep="|")

df.head()
```

```python
print("Shape:", df.shape)
print("Column names:", df.columns.tolist())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))

plt.plot(df["ICULOS"], df["MAP"], marker="o", markersize=4)

plt.xlabel("ICU length of stay (hours)")
plt.ylabel("MAP (mmHg)")
plt.title("Patient p000001: MAP over Time")

plt.grid(alpha=0.3)
plt.show()
```
